```python
import pandas as pd
from pathlib import Path

# Localiza a raiz do projeto a partir do diretorio atual do Jupyter
diretorio_atual = Path.cwd()
raiz_projeto = diretorio_atual

# Sobe os diretorios ate encontrar a pasta 'data'
while raiz_projeto != raiz_projeto.parent:
    if (raiz_projeto / "data").exists():
        break
    raiz_projeto = raiz_projeto.parent

pasta_ibge = raiz_projeto / "data" / "silver" / "ibge"
pasta_caged = raiz_projeto / "data" / "silver" / "caged"

print("Raiz do projeto detectada:", raiz_projeto)
print("Pasta IBGE:", pasta_ibge)
print("Pasta CAGED:", pasta_caged)
print("----------------------------------------")

# 1. Leitura e exibicao dos arquivos do IBGE
if pasta_ibge.exists():
    arquivos_ibge = list(pasta_ibge.glob("*.parquet"))
    print("Arquivos IBGE encontrados:", [f.name for f in arquivos_ibge])

    for arquivo in arquivos_ibge:
        print("\n--- Tabela IBGE:", arquivo.name, "---")
        df_ibge = pd.read_parquet(arquivo)
        print("Formato (linhas, colunas):", df_ibge.shape)
        display(df_ibge.head())
else:
    print("Pasta do IBGE nao encontrada em:", pasta_ibge)

print("----------------------------------------")

# 2. Leitura e exibicao do arquivo do CAGED
caminho_caged = pasta_caged / "caged_2022.parquet"

if caminho_caged.exists():
    print("\n--- Tabela CAGED: caged_2022.parquet ---")
    df_caged = pd.read_parquet(caminho_caged)
    print("Formato (linhas, colunas):", df_caged.shape)

    print("\nTipos das colunas:")
    print(df_caged.dtypes)

    print("\nPrimeiras 5 linhas do CAGED:")
    display(df_caged.head())
else:
    print("Arquivo do CAGED nao encontrado em:", caminho_caged)
```
